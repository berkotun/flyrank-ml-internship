## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is primarily a Scoring and Binary Classification task.
Rather than just categorizing pages, the goal is to assign a "decay risk score" or probability to each page. This allows the SEO team to rank the pages and focus their limited resources on the content that has the highest probability of decaying and the highest potential upside if refreshed.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The exact target is whether a page's organic performance is actively declining.
Since "declining" can be abstract, the concrete proxy we will predict is a binary label: is_declining (e.g., 1 if the trend_direction is 'down' or if traffic has dropped by >20% over the last 90 days, 0 otherwise).

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@K (e.g., Precision@50).
The SEO team can only update a limited number of pages per week (e.g., top 50 recommendations). Therefore, identifying true decay is critical. A high precision ensures that when the model recommends a page for an update, the team doesn't waste hours on a page that was actually healthy (minimizing False Positives).

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

# Load the same anonymized starter data
csv_url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(csv_url)

# Sketching the target column: Creating a binary 'is_declining' label based on trend direction
df['is_declining'] = (df['trend_direction'] == 'down').astype(int)

# The unit of analysis: One row = One specific page/URL evaluated over a recent 90-day window
print("Unit of Analysis: One row represents one page's performance metrics.")
print(f"Dataframe shape: {df.shape}\n")

# Display a sketch of what the model will look at and predict
cols_to_show = ['content_age_days', 'days_since_last_update', 'impressions_90d', 'ctr', 'trend_direction', 'is_declining']
display(df[cols_to_show].head())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule might dictate: "Update any page older than 180 days." However, this heuristic is flawed because some evergreen content remains highly relevant for years without updates, while other fast-moving topics decay in weeks.
ML beats this fixed rule by learning the complex, non-linear interactions between multiple features—such as how content_age_days interacts with avg_position, impressions_90d, and specific content_type. The model uncovers subtle patterns of obsolescence that a human writing static SQL rules would miss.
